```python
from stable_baselines3 import DQN
from stable_baselines3.common.evaluation import evaluate_policy
from socnavgym.wrappers import DiscreteActions
from gymnasium.wrappers import FlattenObservation
from wrapper import *
import socnavgym
import gymnasium as gym
from stable_baselines3.common.monitor import Monitor
import os
```

```python
log_dir = "logs_dqn"

env = gym.make("SocNavGym-v1", config="./environment_configs/my_config.yaml")
env = DiscreteActions(env)

SEED = 1
```

```python
net_arch = [512, 256, 256, 256, 128, 128, 64]
policy_kwargs = {"net_arch" : net_arch}
model = DQN("MultiInputPolicy", env, seed=SEED, verbose=1, policy_kwargs=policy_kwargs, learning_rate=1e-4, tensorboard_log=log_dir)
# Train the agent and display a progress bar
model.learn(total_timesteps=int(1000000), progress_bar=True, tb_log_name="run1", log_interval=1)
# Save the agent
model.save("dqn_lunar")
```

```python
# Load saved model
model = DQN.load("dqn_lunar.zip", env=env)

# Rollout
obs, _ = env.reset(seed=1)
done, truncated = False, False

while not (done or truncated):
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, done, truncated, info = env.step(action)
    env.render()
    if done:
        print(info)
    #time.sleep(0.05)  # control playback speed

env.close()
```
